## Task 1: Regex Fundamentals (Character Classes & Quantifiers)

```python
text1 = "In 2023, Ali bought 3 Laptops for Rs. 45000 each. By 1999 standards, Karachi and Lahore had 12 shops. Email me @ 9am!"
print(text1)
```

```python
# a) Extract all digits (each single digit)
digits = re.findall(r'\d', text1)
print("a) Digits:", digits)

# b) Words starting with a capital letter
cap_words = re.findall(r'\b[A-Z][a-zA-Z]*\b', text1)
print("b) Capitalized words:", cap_words)

# c) Sequences of exactly 4 digits (\b stops it matching inside 45000)
four_digits = re.findall(r'\b\d{4}\b', text1)
print("c) 4-digit sequences:", four_digits)

# d) Words 3 to 6 characters long (letters only)
words_3_6 = re.findall(r'\b[a-zA-Z]{3,6}\b', text1)
print("d) Words of length 3-6:", words_3_6)
```

## Task 2: Pattern-Based Extraction (Real World Data)

```python
text2 = ("""Order ORD-10234 was placed on 15/03/2024 by sara.khan@gmail.com. """
         """Contact her at 0312-4567890 or support at help_desk@shop.com.pk. """
         """A 12.5% discount applied, and 45% of stock was sold by 02/04/2024. """
         """Backup number: 0345-1122334. Order ORD-10235 had 100% refund.""")
print(text2)
```

```python
# a) Email addresses
emails = re.findall(r'[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}', text2)
print("a) Emails:", emails)

# b) Phone numbers in 03XX-XXXXXXX format
phones = re.findall(r'\b03\d{2}-\d{7}\b', text2)
print("b) Phones:", phones)

# c) Dates in DD/MM/YYYY format
dates = re.findall(r'\b(?:0[1-9]|[12]\d|3[01])/(?:0[1-9]|1[0-2])/\d{4}\b', text2)
print("c) Dates:", dates)

# d) Percentages (integer or decimal)
percents = re.findall(r'\d+(?:\.\d+)?%', text2)
print("d) Percentages:", percents)
```

## Task 3: Word Tokenization (NLTK vs spaCy)

```python
para = ("I don't think it's going to rain today, but my mother-in-law said \"take an umbrella.\" "
        "She's a well-known worrier! Email her at aunty.sara@gmail.com if you can't come. Isn't that funny?")

from nltk.tokenize import word_tokenize

# a) NLTK tokenization
nltk_tokens = word_tokenize(para)
print("NLTK tokens:\n", nltk_tokens)

# b) Blank spaCy pipeline (tokenizer only, no trained components)
nlp_blank = spacy.blank("en")
spacy_tokens = [t.text for t in nlp_blank(para)]
print("\nspaCy tokens:\n", spacy_tokens)

# d) Token counts
print("\nNLTK token count :", len(nltk_tokens))
print("spaCy token count:", len(spacy_tokens))
```

```python
# c) Differences observed
print("Only in NLTK :", sorted(set(nltk_tokens) - set(spacy_tokens)))
print("Only in spaCy:", sorted(set(spacy_tokens) - set(nltk_tokens)))

# Difference 1 (hyphenated words): NLTK keeps "mother-in-law" and "well-known" as single tokens,
#   while spaCy splits them on hyphens: "mother", "-", "in", "-", "law".
# Difference 2 (emails): spaCy keeps "aunty.sara@gmail.com" as one token, while NLTK breaks it
#   into "aunty.sara", "@", "gmail.com".
# Difference 3 (quotes): NLTK converts double quotes into `` and '' (Penn Treebank style),
#   while spaCy keeps the original " character.
# Similarity: both split contractions the same way ("do"+"n't", "ca"+"n't", "it"+"'s").
# Result: NLTK = 44 tokens, spaCy = 48 tokens (extra tokens come from hyphen splitting).
```

## Task 4: Token Attributes (Categorizing Tokens)

```python
nlp = spacy.load("en_core_web_sm")

# a) Load sentence into a spaCy Doc
sentence = "Ahmed paid $50 and Rs. 500 for 3 books, then emailed ahmed.raza@example.com twice."
doc = nlp(sentence)

# b) and c) Print is_alpha, like_num, like_email for each token
print(f"{'Token':<25}{'is_alpha':<10}{'like_num':<10}{'like_email':<10}")
for token in doc:
    print(f"{token.text:<25}{str(token.is_alpha):<10}{str(token.like_num):<10}{str(token.like_email):<10}")

# d) Tokens where is_currency is True
currency_tokens = [token.text for token in doc if token.is_currency]
print("\nCurrency tokens:", currency_tokens)
# Note: only "$" is flagged. "Rs" is not in spaCy's currency symbol list, so is_currency is False for it.
```

## Task 5: Sentence Segmentation (Handling Tricky Cases)

```python
# a) Load trained pipeline
nlp = spacy.load("en_core_web_sm")

para5 = ("Yesterday I visited Dr. Ahmed at the city hospital. He told Mr. Khan to avoid junk food, "
         "e.g. chips and soda, for a month. The clinic opens at 9 a.m. on weekdays. "
         "I will go back next Monday.")
doc5 = nlp(para5)

# b) Split into sentences
sentences = list(doc5.sents)
for i, sent in enumerate(sentences, 1):
    print(f"{i}: {sent.text}")

# c) Total sentences
print("\nTotal sentences detected:", len(sentences))
```

```python
# d) Observation: No incorrect break occurred. spaCy correctly treated "Dr.", "Mr.", "e.g." and "a.m."
#    as abbreviations and not sentence endings, so all 4 sentences were detected correctly.
```
